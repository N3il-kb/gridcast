```python
import geopandas as gpd
from shapely.geometry import Polygon
import numpy as np

# 1. Load USA polygon
usa = gpd.read_file(
    "https://raw.githubusercontent.com/PublicaMundi/MappingAPI/master/data/geojson/us-states.json"
)
usa_border = usa.unary_union
usa_gdf = gpd.GeoDataFrame(geometry=[usa_border], crs="EPSG:4326")

# 2. Reproject to Albers Equal Area (CRS suitable for USA)
usa_proj = usa_gdf.to_crs("EPSG:5070")  # NAD83 / Conus Albers

# 3. Hex radius in meters (distance from center to vertex)
HEX_RADIUS = 50000  # 50 km radius hex

def make_hex(center_x, center_y, radius):
    """Return a perfect pointy-top hexagon polygon with tiny overlap."""
    angles = np.radians(np.arange(30, 390, 60))
    # Add 0.1% overlap to ensure no gaps due to floating point precision
    overlap_factor = 1.001
    return Polygon([
        (center_x + radius * overlap_factor * np.cos(a), 
         center_y + radius * overlap_factor * np.sin(a))
        for a in angles
    ])

# --- Correct tiling spacing ---
hex_width = 2 * HEX_RADIUS                        # flat-to-flat width
hex_height = np.sqrt(3) * HEX_RADIUS              # point-to-point height

dx = np.sqrt(3) * HEX_RADIUS                             # horizontal spacing
dy = 0.865 * hex_height                                    # vertical spacing

# 4. Build bounding box in projected coordinates
minx, miny, maxx, maxy = usa_proj.total_bounds

# 5. Generate hex centers
hexes = []
row = 0
y = miny - hex_height

while y < maxy + hex_height:
    # offset every other row
    x_offset = (row % 2) * (dx / 2)  # = 0.75 * HEX_RADIUS
    x = minx - hex_width
    while x < maxx + hex_width:
        hexes.append(make_hex(x + x_offset, y, HEX_RADIUS))
        x += dx
    y += dy
    row += 1

hexgrid_proj = gpd.GeoDataFrame(geometry=hexes, crs="EPSG:5070")

# 6. Clip hexgrid to US border
hex_us_proj = gpd.overlay(hexgrid_proj, usa_proj, how="intersection")

# 7. Reproject back to WGS84 for Folium / display
hex_us = hex_us_proj.to_crs("EPSG:4326")

# 8. Visualize (geopandas)
hex_us.explore(color="green")
```

```python
import pandas as pd
import geopandas as gpd
from sklearn.neighbors import NearestNeighbors

# --- STEP 1: Load Real Data & Make it Spatial ---
# Load the CSV you just created
df_real = pd.read_csv('datacenter_scores_real.csv')

# Convert to a GeoDataFrame (Points) using Lat/Lon
gdf_scores = gpd.GeoDataFrame(
    df_real,
    geometry=gpd.points_from_xy(df_real.lon, df_real.lat),
    crs="EPSG:4326"
)

# Project to match your Hex Grid (EPSG:5070 / Albers)
# We assume 'hex_us_proj' exists from your previous cell
gdf_scores_proj = gdf_scores.to_crs(hex_us_proj.crs)

# --- STEP 2: Broadcast Data to Grid (The "Merge") ---
# This command finds the nearest region point for every single hexagon 
# and copies the scores (dc_score, profitability, etc.) to that hex.
hex_scored = gpd.sjoin_nearest(
    hex_us_proj,
    gdf_scores_proj,
    how="left",
    distance_col="dist_to_region"
)

# Cleanup: sjoin_nearest adds an 'index_right' column we don't need
hex_scored = hex_scored.drop(columns=['index_right']).reset_index(drop=True)

# Stable ID per hex for joining other datasets (e.g., weather)
hex_scored['hex_id'] = np.arange(len(hex_scored))
```

```python
import folium

# --- STEP 3: Fold temperature into profitability (10% of total DC score) and smooth ---
hex_temp_df = pd.read_csv('hex_weather_data_all.csv')
hex_scored = hex_scored.merge(hex_temp_df, on="hex_id", how="left")

# Normalize temperature so cooler locations score higher
temp_min, temp_max = hex_scored["local_temp_c"].min(), hex_scored["local_temp_c"].max()
temp_span = temp_max - temp_min if temp_max != temp_min else 1
hex_scored["temp_cool_score"] = 1 - ((hex_scored["local_temp_c"] - temp_min) / temp_span)
hex_scored["temp_cool_score"] = hex_scored["temp_cool_score"].clip(0, 1)
hex_scored["temp_cool_score"] = hex_scored["temp_cool_score"].fillna(hex_scored["temp_cool_score"].mean())

# Profitability with temperature baked in; temp is 25% of profitability so it's 10% of total DC (0.25 * 0.40 = 0.10)
hex_scored["profitability_temp_adj"] = (
    0.75 * hex_scored["profitability"] +
    0.25 * hex_scored["temp_cool_score"]
)

# Recompute DC score with the adjusted profitability
hex_scored["dc_score_temp"] = (
    0.40 * hex_scored["profitability_temp_adj"] +
    0.60 * hex_scored["sustainability"]
)

# --- Visual smoothing ---
centers = np.vstack(hex_scored.geometry.centroid.apply(lambda p: (p.x, p.y)))
nbrs = NearestNeighbors(n_neighbors=15).fit(centers)
d, idx = nbrs.kneighbors(centers)
smooth_score = np.array([hex_scored["dc_score_temp"].iloc[i].mean() for i in idx])
hex_scored["dc_score_smooth"] = smooth_score

# --- STEP 4: Visualize (three layers) ---
hex_map_final = hex_scored.to_crs("EPSG:4326")

# Temperature layer
score_map = hex_map_final.explore(
    column="local_temp_c",
    cmap="OrRd",
    tiles="CartoDB positron",
    tooltip=["hex_id", "local_temp_c"],
    style_kwds={
        "fillOpacity": 0.55,
        "weight": 0,
        "opacity": 0
    },
    legend_kwds={"caption": "Temperature (°C)"},
    name="Temperature (°C)",
)

# Elevation layer
hex_map_final.explore(
    column="elevation_m",
    cmap="terrain",
    m=score_map,
    tiles=None,
    tooltip=["hex_id", "elevation_m"],
    style_kwds={
        "fillOpacity": 0.55,
        "weight": 0,
        "opacity": 0
    },
    legend_kwds={"caption": "Elevation (m)"},
    name="Elevation (m)",
)

# Datacenter score layer (added last so its tooltip is default)
hex_map_final.explore(
    column="dc_score_smooth",
    cmap="BuGn",
    m=score_map,
    tiles=None,
    tooltip=[
        "region",
        "dc_score_temp",
        "dc_score_smooth",
        "profitability_temp_adj",
        "sustainability",
        "local_temp_c",
        "elevation_m",
    ],
    style_kwds={
        "fillOpacity": 0.75,
        "weight": 0,
        "opacity": 0
    },
    legend_kwds={"caption": "GridScore"},
    name="Datacenter score",
)

score_map.fit_bounds([[24.5, -124.7], [49.5, -66.9]])

folium.LayerControl(collapsed=False).add_to(score_map)
score_map
```
